```python
import json

# STEP 1 & 2: Environment Setup & Knowledge Base Creation
# We create the file programmatically to ensure Step 2 is complete.
kb_data = {
    "network_slow": {"symptom": "slow internet", "solution": "Restart your router and check cable connections."},
    "software_freeze": {"symptom": "software keeps crashing", "solution": "Update the app to the latest version and restart your PC."},
    "system_heat": {"symptom": "computer is hot", "solution": "Clear dust from vents and close background apps."},
    "browser_issue": {"symptom": "browser cache", "solution": "Clear browser cookies and history."}
}

with open('troubleshooting_knowledge_base.json', 'w') as f:
    json.dump(kb_data, f)

# STEP 4 & 5: Diagnostic Logic & Automated Fixes
def automate_system_refresh():
    print("\n[🤖 Automation] Clearing temporary system files...")
    print("[🤖 Automation] Flushing DNS cache...")
    print("✅ Automated maintenance complete.")

def run_diagnostics(user_input):
    # Specialized logic for Network (Step 4)
    if "internet" in user_input or "wifi" in user_input:
        print("\n🔍 Agent: I've detected a network-related issue.")
        cables = input("Agent: Are your cables plugged in securely? (Yes/No): ").lower()
        if cables == "no":
            return "Please plug in your cables and try again."
        else:
            automate_system_refresh() # Step 5: Automation
            return "I have refreshed your network settings. Please check your speed now."
    return None

# STEP 3 & 6: User Interaction & Testing
def start_intelligent_agent():
    # Load KB
    with open('troubleshooting_knowledge_base.json', 'r') as f:
        kb = json.load(f)

    print("--- 🧠 Intelligent Troubleshooting Agent Online ---")
    user_problem = input("Agent: Please describe the technical issue you are facing:\nUser: ").lower()

    # 1. Try Diagnostic Logic first
    recommendation = run_diagnostics(user_problem)

    # 2. Fallback to Knowledge Base Search
    if not recommendation:
        for key, data in kb.items():
            if data["symptom"] in user_problem:
                recommendation = data["solution"]
                break

    # 3. Output results
    if recommendation:
        print(f"\nAgent: {recommendation}")
    else:
        print("\nAgent: I'm not sure how to fix that yet. Try checking your user manual.")

# Execute Step 6 (Testing)
start_intelligent_agent()
```
